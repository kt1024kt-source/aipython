```python
import pandas as pd
#산지니로 예제 만들어보기
# 1) 예시 데이터 만들기(보통은 CSV/DB에서 읽어오지만, 여기선 바로 생성)
data = [
    {"date": "2026-05-01", "store": "Seoul",  "category": "Coffee", "units": 12, "unit_price": 4.5},
    {"date": "2026-05-01", "store": "Seoul",  "category": "Tea",    "units":  8, "unit_price": 3.8},
    {"date": "2026-05-01", "store": "Busan",  "category": "Coffee", "units": 10, "unit_price": 4.3},
    {"date": "2026-05-02", "store": "Seoul",  "category": "Coffee", "units": 15, "unit_price": 4.6},
    {"date": "2026-05-02", "store": "Busan",  "category": "Tea",    "units":  7, "unit_price": 3.7},
    {"date": "2026-05-03", "store": "Busan",  "category": "Coffee", "units":  9, "unit_price": 4.4},
]

df = pd.DataFrame(data)

# 2) 타입 변환(날짜 문자열 -> datetime)
df["date"] = pd.to_datetime(df["date"])

# 3) 파생 컬럼 만들기 (매출 = 수량 * 단가)
df["revenue"] = df["units"] * df["unit_price"]

print("=== 원본 데이터 ===")
print(df)

# 4) 필터링: 서울 매장 + 커피만
seoul_coffee = df[(df["store"] == "Seoul") & (df["category"] == "Coffee")]
print("\n=== 필터링(Seoul & Coffee) ===")
print(seoul_coffee)

# 5) 그룹 집계: 매장/카테고리별 총 판매수량, 총매출, 평균단가
summary = (
    df.groupby(["store", "category"])
      .agg(total_units=("units", "sum"),
           total_revenue=("revenue", "sum"),
           avg_unit_price=("unit_price", "mean"))
      .reset_index()
      .sort_values("total_revenue", ascending=False)
)

print("\n=== 그룹 집계 결과(store, category) ===")
print(summary)

# 6) 피벗 테이블: 날짜별 매장 총매출(가로로 펼치기)
pivot = df.pivot_table(
    index="date",
    columns="store",
    values="revenue",
    aggfunc="sum",
    fill_value=0
)

print("\n=== 피벗(날짜별 매장 매출) ===")
print(pivot)

# 7) 결측치 처리 맛보기: 일부 값을 NaN으로 만들고 채우기
df2 = df.copy()
df2.loc[0, "unit_price"] = None

print("\n=== 결측치 포함 ===")
print(df2[["date", "store", "category", "units", "unit_price", "revenue"]])

# unit_price 결측치를 매장별 평균 단가로 채우기
df2["unit_price"] = df2["unit_price"].fillna(df2.groupby("store")["unit_price"].transform("mean"))
df2["revenue"] = df2["units"] * df2["unit_price"]

print("\n=== 결측치 처리 후(매장별 평균으로 채움) ===")
print(df2[["date", "store", "category", "units", "unit_price", "revenue"]])
```
